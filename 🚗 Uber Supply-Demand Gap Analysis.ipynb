## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# ═══════════════════════════════════════════════════════
# 3a. Parse mixed-format timestamps
# Two formats found in the data:
#   "11/7/2016 11:51"  → day/month/year HH:MM
#   "13-07-2016 08:33:16" → day-month-year HH:MM:SS
# ═══════════════════════════════════════════════════════

def parse_ts(ts_series):
    """Parse a mixed-format timestamp series robustly."""
    out = pd.to_datetime(ts_series, dayfirst=True, errors='coerce')
    mask = out.isna()
    if mask.any():
        out[mask] = pd.to_datetime(ts_series[mask], format='%d-%m-%Y %H:%M:%S',
                                   errors='coerce')
    return out

df = df_raw.copy()
df['Request_ts']  = parse_ts(df['Request timestamp'])
df['Drop_ts']     = parse_ts(df['Drop timestamp'])

print("Timestamp parse success rate:")
print(f"  Request_ts : {df['Request_ts'].notna().sum()} / {len(df)}")
print(f"  Drop_ts    : {df['Drop_ts'].notna().sum()} / {len(df)}")
````

````python
# ═══════════════════════════════════════════════════════
# 3b. Feature Engineering
# ═══════════════════════════════════════════════════════
df['Hour']    = df['Request_ts'].dt.hour
df['Day']     = df['Request_ts'].dt.day
df['Weekday'] = df['Request_ts'].dt.day_name()

# Time-of-day buckets matching the PDF analysis
def time_slot(h):
    if   0 <= h < 5:  return 'Early Morning'
    elif 5 <= h < 10: return 'Morning'
    elif 10 <= h < 14: return 'Afternoon'
    elif 14 <= h < 18: return 'Evening'
    elif 18 <= h < 21: return 'Late Night'
    else:              return 'Night'

df['Time_slot'] = df['Hour'].apply(time_slot)

# Trip duration (minutes) for completed trips
df['Duration_min'] = (df['Drop_ts'] - df['Request_ts']).dt.total_seconds() / 60

# Is Unfulfilled?
df['Unfulfilled'] = df['Status'].isin(['Cancelled', 'No Cars Available']).astype(int)

print("Feature engineering complete. New columns:")
print(df[['Hour','Day','Weekday','Time_slot','Duration_min','Unfulfilled']].head(8))
````

````python
# ═══════════════════════════════════════════════════════
# 3c. Rename columns for cleanliness
# ═══════════════════════════════════════════════════════
df.rename(columns={'Pickup point': 'Pickup_point',
                   'Driver id'   : 'Driver_id',
                   'Request id'  : 'Request_id'}, inplace=True)

print("Final cleaned dataset shape:", df.shape)
df.head()
````

# 5. SQL — Insight Queries (SQLite)

> Load the cleaned data into an in-memory SQLite database and run structured queries
> that mirror what a data analyst would write in any SQL environment.


````python
# ═══════════════════════════════════════════════════════════
# SQL STEP 1 — Create SQLite in-memory DB and load data
# ═══════════════════════════════════════════════════════════

conn = sqlite3.connect(':memory:')   # In-memory DB — no file needed

# Write the cleaned DataFrame as a SQL table
df.to_sql('uber', conn, if_exists='replace', index=False)
print("✅ Table 'uber' loaded into SQLite.")
print(f"   Rows: {pd.read_sql('SELECT COUNT(*) AS cnt FROM uber', conn).iloc[0,0]:,}")
````

````python
# ─── SQL QUERY 1: Overall Status Distribution ───────────────
q1 = """
SELECT  Status,
        COUNT(*)                               AS Total_Requests,
        ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM uber), 2) AS Pct
FROM    uber
GROUP BY Status
ORDER BY Total_Requests DESC;
"""
sql_q1 = pd.read_sql(q1, conn)
print("── Q1: Overall Status Distribution ──")
print(sql_q1.to_string(index=False))
````

````python
# ─── SQL QUERY 2: Supply-Demand Gap by Time Slot ────────────
q2 = """
SELECT  Time_slot,
        COUNT(*)                               AS Total,
        SUM(Unfulfilled)                       AS Unfulfilled,
        ROUND(SUM(Unfulfilled)*100.0/COUNT(*), 2) AS Unfulfilled_Pct
FROM    uber
GROUP BY Time_slot
ORDER BY Unfulfilled DESC;
"""
sql_q2 = pd.read_sql(q2, conn)
print("── Q2: Gap by Time Slot ──")
print(sql_q2.to_string(index=False))
````

````python
# ─── SQL QUERY 3: Cancellation Rate by Pickup Point ─────────
q3 = """
SELECT  Pickup_point,
        COUNT(*)                                   AS Total_Requests,
        SUM(CASE WHEN Status='Cancelled' THEN 1 ELSE 0 END) AS Cancelled,
        SUM(CASE WHEN Status='No Cars Available' THEN 1 ELSE 0 END) AS No_Cars,
        ROUND(SUM(CASE WHEN Status='Cancelled' THEN 1.0 ELSE 0 END)/COUNT(*)*100,2)
            AS Cancel_Pct,
        ROUND(SUM(CASE WHEN Status='No Cars Available' THEN 1.0 ELSE 0 END)/COUNT(*)*100,2)
            AS NoCar_Pct
FROM    uber
GROUP BY Pickup_point;
"""
sql_q3 = pd.read_sql(q3, conn)
print("── Q3: Cancellation & No Car Rate by Pickup Point ──")
print(sql_q3.to_string(index=False))
````

````python
# ─── SQL QUERY 4: Peak Gap Hours ────────────────────────────
q4 = """
SELECT  Hour,
        COUNT(*)        AS Requests,
        SUM(Unfulfilled) AS Unfulfilled,
        ROUND(AVG(Unfulfilled)*100, 2) AS Gap_Pct
FROM    uber
GROUP BY Hour
ORDER BY Unfulfilled DESC
LIMIT 10;
"""
sql_q4 = pd.read_sql(q4, conn)
print("── Q4: Top-10 Gap Hours ──")
print(sql_q4.to_string(index=False))
````

````python
# ─── SQL QUERY 5: Average Trip Duration by Pickup Point ─────
q5 = """
SELECT  Pickup_point,
        ROUND(AVG(Duration_min), 2)  AS Avg_Duration_Min,
        ROUND(MIN(Duration_min), 2)  AS Min_Duration_Min,
        ROUND(MAX(Duration_min), 2)  AS Max_Duration_Min,
        COUNT(*)                     AS Completed_Trips
FROM    uber
WHERE   Status = 'Trip Completed'
GROUP BY Pickup_point;
"""
sql_q5 = pd.read_sql(q5, conn)
print("── Q5: Trip Duration Stats by Pickup Point ──")
print(sql_q5.to_string(index=False))
````

````python
# ─── SQL QUERY 6: Daily Trend ───────────────────────────────
q6 = """
SELECT  Day,
        COUNT(*)        AS Total,
        SUM(CASE WHEN Status='Trip Completed' THEN 1 ELSE 0 END)      AS Completed,
        SUM(CASE WHEN Status='Cancelled' THEN 1 ELSE 0 END)           AS Cancelled,
        SUM(CASE WHEN Status='No Cars Available' THEN 1 ELSE 0 END)   AS No_Cars
FROM    uber
GROUP BY Day
ORDER BY Day;
"""
sql_q6 = pd.read_sql(q6, conn)
print("── Q6: Daily Request Trend ──")
print(sql_q6.to_string(index=False))
````

````python
# ─── SQL QUERY 7: Worst Pickup × Time-Slot Combos ───────────
q7 = """
SELECT  Pickup_point, Time_slot,
        COUNT(*)         AS Total,
        SUM(Unfulfilled) AS Unfulfilled,
        ROUND(SUM(Unfulfilled)*100.0/COUNT(*),1) AS Gap_Pct
FROM    uber
GROUP BY Pickup_point, Time_slot
ORDER BY Unfulfilled DESC
LIMIT 10;
"""
sql_q7 = pd.read_sql(q7, conn)
print("── Q7: Worst Pickup × Time-Slot Combinations ──")
print(sql_q7.to_string(index=False))
````

#### Chart - 1 — Status Distribution (Univariate · Categorical)

````python
# Chart - 1 visualization code

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

status_counts = df['Status'].value_counts()
colors = ['#2ECC71', '#E74C3C', '#E67E22']

# Bar
axes[0].bar(status_counts.index, status_counts.values, color=colors, edgecolor='white', linewidth=1.3)
axes[0].set_title('Ride Request Status — Count', fontweight='bold')
axes[0].set_ylabel('Number of Requests')
for i, v in enumerate(status_counts.values):
    axes[0].text(i, v + 30, str(v), ha='center', fontweight='bold')

# Pie
wedges, texts, autotexts = axes[1].pie(
    status_counts.values, labels=status_counts.index,
    autopct='%1.1f%%', colors=colors, startangle=140,
    wedgeprops=dict(edgecolor='white', linewidth=1.5))
[t.set_fontsize(11) for t in autotexts]
axes[1].set_title('Ride Request Status — Share', fontweight='bold')

plt.suptitle('Fig 1 | Overall Request Status Distribution', fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout(); plt.show()
````

##### 1. Why did you pick the specific chart?

Answer Here.

- A dual-panel bar + pie/donut layout was chosen because Status is a nominal categorical variable with only three mutually exclusive levels — a textbook scenario for univariate categorical analysis. The two panels serve complementary purposes:
  - The bar chart encodes absolute frequency (raw counts), making it easy to compare magnitudes directly — you can see that "Trip Completed" is almost exactly 2× "No Cars Available".
  - The donut chart encodes relative frequency (proportional share), giving the part-to-whole picture at a glance.
    
- Using both panels together avoids a common analytical blind spot: a pie alone hides scale (is this 100 trips or 100,000?), while a bar alone buries the proportional story. Together they answer both "how many?" and "how much of the total?" in a single figure, which is exactly what a business stakeholder needs.



##### 2. What is/are the insight(s) found from the chart?

Answer Here

**Insight 1 — Bare majority of trips complete.** Only 53.5% of all ride requests result in a completed trip. In a ride-hailing business this is a critically low fulfillment rate — a healthy platform typically aims for 70–80%+ completion rates.

**Insight 2 — Supply-side failure dominates unfulfilled demand.** "No Cars Available" accounts for 23.9% of all requests — nearly one in four users who open the app get nothing because supply is simply absent. This is a supply-side gap, not a user behaviour problem.

**Insight 3 — Demand-side cancellations are nearly equal to supply failures.** "Cancelled" is 22.7%, extremely close to "No Cars Available" (23.9%). Together these two failure modes account for 46.6% — nearly half of all requests — a striking number for a service business.

**Insight 4 — The two failure modes are near-symmetrical.** The ratio of No Cars : Cancelled is ~1.05:1, suggesting that even when cars do exist, nearly as many rides collapse from the demand side (user or driver cancellation) as from the supply side. This points to a dual problem, not a single root cause.

#### Chart - 4 — Trip Duration Distribution (Univariate · Numerical)

````python
# Chart - 4 visualization code
completed = df[df['Status'] == 'Trip Completed']['Duration_min'].dropna()
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].hist(completed, bins=30, color='#27AE60', edgecolor='white', linewidth=1.2)
axes[0].axvline(completed.mean(), color='red', linestyle='--', label=f'Mean: {completed.mean():.1f} min')
axes[0].set_title('Trip Duration — Histogram', fontweight='bold')
axes[0].set_xlabel('Minutes'); axes[0].legend()

axes[1].boxplot(completed, vert=False, patch_artist=True,
                boxprops=dict(facecolor='#27AE60', color='#1A5E35'),
                medianprops=dict(color='red', linewidth=2))
axes[1].set_title('Trip Duration — Box Plot', fontweight='bold')
axes[1].set_xlabel('Minutes')

plt.suptitle('Fig 4 | Completed Trip Duration Distribution', fontweight='bold', y=1.01)
plt.tight_layout(); plt.show()
print(f"Mean: {completed.mean():.1f} min | Median: {completed.median():.1f} min | Max: {completed.max():.1f} min")
````

Thirty bins give fine enough resolution to detect the distribution's peak and the gradual right tail, without over-fragmenting into noise. The superimposed mean line as a dashed vertical immediately communicates skew direction: when the mean sits visibly to the right of the modal bin, the distribution is positively skewed.

The **box plot** answers spread and outliers — Q1, median, Q3, IQR, whisker extent, and individual outlier points that the histogram's binning can obscure. Together the two panels answer all five key distribution questions (centre, spread, shape, outliers, range) with no redundancy — each panel contributes information the other cannot show clearly.

Filtering to **completed trips only** is the correct analytical decision: including cancelled or no-car-available records would introduce zeros or nulls into duration, masking the true service time distribution.

The fill=True, alpha=0.3 transparency allows all three distributions to coexist in the same panel without occlusion, making overlap zones immediately visible as color blends. #### Chart 20 — Final: Supply vs Demand by Time Slot (Multivariate · Grouped Bar)

````python
slot_order = ['Early Morning','Morning','Afternoon','Evening','Late Night','Night']
demand = df.groupby('Time_slot').size().reindex(slot_order)
supply = df[df['Status']=='Trip Completed'].groupby('Time_slot').size().reindex(slot_order)

fig, ax = plt.subplots(figsize=(12, 6))
x = np.arange(len(slot_order)); w = 0.35
b1 = ax.bar(x - w/2, demand, w, label='Demand (Total Requests)', color='#3498DB', edgecolor='white')
b2 = ax.bar(x + w/2, supply, w, label='Supply (Completed Trips)', color='#27AE60', edgecolor='white')

# Gap annotation arrows
for xi, (d, s) in enumerate(zip(demand, supply)):
    gap = d - s
    ax.annotate(f'Gap {gap:,}', xy=(xi + w/2, s), xytext=(xi, s + gap/2),
                arrowprops=dict(arrowstyle='->', color='red', lw=1.5),
                fontsize=8, color='red', ha='center')

ax.set_xticks(x); ax.set_xticklabels(slot_order, rotation=20, ha='right')
ax.set_title('Fig 20 | 📊 Supply vs Demand Gap — By Time Slot', fontweight='bold', fontsize=15)
ax.set_ylabel('Number of Requests')
ax.legend(fontsize=11)
plt.tight_layout(); plt.show()
print("✅ All 20 charts complete!")
````
